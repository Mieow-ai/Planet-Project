```python
ua = 149597887.5

for planete in systeme_solaire.keys():
    systeme_solaire[planete]["c"] = systeme_solaire[planete]["demi_grand_axe_km"] - systeme_solaire[planete]["périhélie_km"]
    systeme_solaire[planete]["demi_g_axe_ua"] = systeme_solaire[planete]["demi_grand_axe_km"] / ua
    systeme_solaire[planete]["c_ua"] = systeme_solaire[planete]["c"] / ua
    systeme_solaire[planete]["demi_p_axe_ua"]= sqrt(systeme_solaire[planete]["demi_g_axe_ua"]**2 - systeme_solaire[planete]["c_ua"]**2)

for satellite in satellites.keys():
    satellites[satellite]["c"] = satellites[satellite]["demi_grand_axe_km"] - satellites[satellite]["périgée_km"]
    satellites[satellite]["demi_g_axe_ua"] = satellites[satellite]["demi_grand_axe_km"] / ua
    satellites[satellite]["c_ua"] = satellites[satellite]["c"] / ua
    satellites[satellite]["demi_p_axe_ua"] = sqrt(satellites[satellite]["demi_g_axe_ua"]**2 - satellites[satellite]["c_ua"]**2)
```

```python
soleil_rayon_ua = 0.004649
ua = 149597887.5

    
def dessine_planetes_3d_ellipse(k):
    S = sphere(center=(0, 0, 0), size = 30, opacity = 0)
    for planete in systeme_solaire.keys():
        ang = (2*pi / systeme_solaire[planete]["période"])*3*k # Le 3 permet d'accélérer et de devoir produire moins de modélisation
        demi_g_axe = systeme_solaire[planete]["demi_g_axe_ua"]
        demi_p_axe = systeme_solaire[planete]["demi_p_axe_ua"]
        inclinaison =  systeme_solaire[planete]["inclinaison"]*pi/180 # On converti l'inclinaison en radiant
        x =  -systeme_solaire[planete]["c_ua"] + demi_g_axe * cos(ang)
        y = demi_p_axe * sin(ang) * cos(inclinaison)
        z = demi_p_axe * sin(ang) * sin(inclinaison)
        rayon_ua = 1000*systeme_solaire[planete]["rayon_km"]/ua # On multiplie par 1000 les rayon des planetes
        S += sphere((x,y,z), size=rayon_ua, color=systeme_solaire[planete]["couleur"])
        if planete == "Terre":
            ang = (2*pi / satellites["Lune"]["période"])*3*k # Le 3 permet d'accélérer et de devoir produire moins de modélisation
            demi_g_axe = satellites["Lune"]["demi_g_axe_ua"]
            demi_p_axe = satellites["Lune"]["demi_p_axe_ua"]
            inclinaison =  satellites["Lune"]["inclinaison"]*pi/180 # On converti l'inclinaison en radiant
            # Coordonnées de la Lune par rapport à la Terre
            xl_t =  -satellites["Lune"]["c_ua"] + demi_g_axe * cos(ang) 
            yl_t = demi_p_axe * sin(ang) * cos(inclinaison)
            zl_t = demi_p_axe * sin(ang) * sin(inclinaison)
            # Coordonnées de la Lune par rapport au Soleil (origine)
            # On multiplie par 23 la distance entre la Lune et la Terre car la taille des deux astres est multipliée par 1000 et ils se chevauchent
            xl = 23*xl_t + x
            yl = 23*yl_t + y
            zl = 23*zl_t + z
            rayon_ua = 1000*satellites["Lune"]["rayon_km"]/ua # On multiplie par 1000 le rayon de la Lune
            S += sphere((xl,yl,zl), size=rayon_ua, color=satellites["Lune"]["couleur"])
            
    return S

    
def animation_systeme_solaire(N):
    # On modélise une seule fois les ellipses et le Soleil
    S = sphere(center=(0, 0, 0), size = 30, opacity = 0)
    S += sphere(center=(0, 0, 0), size = soleil_rayon_ua*10, opacity = 100, color="orange") # On multiplie par 10 le rayon du soleil pour le voir

    for planete in systeme_solaire.keys(): 
        demi_g_axe = systeme_solaire[planete]["demi_g_axe_ua"]
        demi_p_axe = systeme_solaire[planete]["demi_p_axe_ua"]
        inclinaison_rad =  systeme_solaire[planete]["inclinaison"]*pi/180 # On converti l'inclinaison en radiant
        t = var("t")
        x = -systeme_solaire[planete]["c_ua"] + demi_g_axe * cos(t)
        y = demi_p_axe * sin(t) * cos(inclinaison_rad)
        z = demi_p_axe * sin(t) * sin(inclinaison_rad)
        S += parametric_plot3d((x, y, z), (t, 0, 2*pi), color="black") 
    liste = [ S + dessine_planetes_3d_ellipse(k) for k in range(1, N)]    # On ajoute les planetes bien positionnées
    
    return animate(liste, figsize=(25, 25))
anim = animation_systeme_solaire(150)
```

# Envoi de la sonde (Modélisation 3d - circulaire)

```python
systeme_solaire2 = {
    "Mercure": {
        "distance_ua": 0.4,
        "période": 87.969,
        "inclinaison": 7.005,
        "couleur": "gray",
        "rayon_km": 2439.7
    },
    "Vénus": {
        "distance_ua": 0.7,
        "période": 224.701,
        "inclinaison": 3.394,
        "couleur": "white",
        "rayon_km": 6051.8
    },
    "Terre": {
        "distance_ua": 1.0,
        "période": 365.256,
        "inclinaison": 0.000,
        "couleur": "blue",
        "rayon_km": 6378.1
    },
    "Mars": {
        "distance_ua": 1.5,
        "période": 686.980,
        "inclinaison": 1.850,
        "couleur": "red",
        "rayon_km": 3396.2
    },
}
satellites = {
    "Lune": {
        "distance_ua":0.00257,
        "période": 27.32,
        "inclinaison": 5.145,
        "couleur": "grey",
        "rayon_km": 1737.4
    }
}
soleil_rayon_ua = 0.004649
ua = 149597887.5
```

```python
rayon_ua_f = 0.01
def dessine_planetes_3d_sonde(k):
    S = sphere(center=(0, 0, 0), size = 0.01, opacity = 0)
    for nom_planete, dic in systeme_solaire2.items():
            
        
        angle = (2*pi / dic["période"])*3*k # Le 3 permet d'accélérer et de devoir produire moins de modélisation

        x = dic["distance_ua"] * cos(angle)
        y = dic["distance_ua"] * sin(angle)
        rayon_ua = 1500*dic["rayon_km"]/ua # On multiplie par 1500 les rayon des planetes    
        S += sphere(center=(x, y, 0), size=rayon_ua, fill=True, color= dic["couleur"])
        if nom_planete == "Terre" :
            angle = (2*pi / satellites["Lune"]["période"])*3*k # Le 3 permet d'accélérer et de devoir produire moins de modélisation
            xl_t = satellites["Lune"]["distance_ua"] * cos(angle)
            yl_t = satellites["Lune"]["distance_ua"] * sin(angle)
            xl = 36*xl_t + x
            yl = 36*yl_t + y 
            rayon_ua = 1500*satellites["Lune"]["rayon_km"]/ua # On multiplie par 1500 le rayon de la Lune pour garder à l'échelle 
            S += sphere(center=(xl, yl, 0), size=rayon_ua, fill=True, color="grey")
            if k >= 175 and k < 221:

                distance_m = systeme_solaire2["Mars"]["distance_ua"]
                angle_m = (2 * pi /systeme_solaire2["Mars"]["période"])* 3 * k
                xm = distance_m * cos(angle_m)
                ym = distance_m * sin(angle_m)
                q = k % 175 + 1
                #FUSÉE
                direction_x = xm - x
                direction_y = ym - y
                direction_z = 0
                # Normalisation du vecteur direction
                norme = sqrt(direction_x**2 + direction_y**2 + direction_z**2)
                dir_x = direction_x / norme
                dir_y = direction_y / norme
                dir_z = direction_z / norme
    
                # Position de la fusée après un certain "temps"
                xf = x + dir_x * q / 40
                yf = y + dir_y * q / 40
                zf =  dir_z * q / 40
                
                S += sphere((xf, yf, zf), size=rayon_ua_f, fill=True, color='red')
        if nom_planete == "Mars" :
            if k >= 221:
                xf2 = x + 0.05 * cos(3*k*pi/(25))
                yf2 = y + 0.05 * sin(3*k*pi/(25))
                zf2 = 0
                S += sphere((xf2, yf2, zf2), size=rayon_ua_f, fill=True, color='red')
    return S

# Créer l'animation
def animation_systeme_solaire_v2():
    # On modélise une seule fois les trajectoires et le Soleil
    S = sphere(center=(0, 0, 0), size = 1, opacity = 0)
    S += sphere(center=(0, 0, 0), size = soleil_rayon_ua*15, opacity = 100, color="orange") # On multiplie par 15 le rayon du soleil pour le voir

    for planete in systeme_solaire2.keys(): 
        S += circle(center=(0, 0), radius=systeme_solaire2[planete]["distance_ua"], fill=False, color='black')
    liste = [S+dessine_planetes_3d_sonde(k) for k in range(150, 280)]
    return animate(liste, figsize=(25, 25))
```
